```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread(r"E:\SRM\Digital Image\Lab 7 - 28th October\Cameraman.png",0)
M,N = img.shape
```

### Develop programs to implement Gaussian, uniform and impulse noise.

```python
gauss_noise=np.zeros((M,N),dtype=np.uint8)
cv2.randn(gauss_noise,128,20)
gauss_noise=(gauss_noise*0.5).astype(np.uint8)

# uni_noise=np.zeros((640,480),dtype=np.uint8)
uni_noise=np.zeros((M,N),dtype=np.uint8)
cv2.randu(uni_noise,0,255)
uni_noise=(uni_noise*0.5).astype(np.uint8)

# imp_noise=np.zeros((640,480),dtype=np.uint8)
imp_noise=np.zeros((M,N),dtype=np.uint8)
cv2.randu(imp_noise,0,255)
imp_noise=cv2.threshold(imp_noise,200,255,cv2.THRESH_BINARY)[1]
```

### Add this to an image.

```python
img1 = img + gauss_noise
img2 = img + uni_noise
img3 = img + imp_noise
```

```python

plt.imshow(img1,cmap='gray')
plt.title('Gaussian Noise')
plt.axis("off")
```

```python

plt.imshow(img2,cmap='gray')
plt.title('Uniform Noise')
plt.axis("off")
```

```python

plt.imshow(img3,cmap='gray')
plt.title('Impulse Noise')
plt.axis("off")
```

### Apply denoising using mean, median and gaussian filters in spatial domain. Compare using MSE.

```python
mean_filtered = cv2.blur(img1, (5, 5))           # Mean filter
median_filtered = cv2.medianBlur(img1, 5)        # Median filter
gaussian_filtered = cv2.GaussianBlur(img1, (5, 5), 0)  # Gaussian filter

# Function to calculate Mean Squared Error
def mse(original, denoised):
    return np.mean((original - denoised) ** 2)

# Calculate MSE for each filtered image
mse_mean = mse(img, mean_filtered)
mse_median = mse(img, median_filtered)
mse_gaussian = mse(img, gaussian_filtered)
```

```python
plt.subplot(3, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis("off")

plt.subplot(3, 2, 2)
plt.imshow(img1, cmap='gray')
plt.title('Image with Gaussian Noise')
plt.axis("off")

plt.subplot(3, 2, 3)
plt.imshow(mean_filtered, cmap='gray')
plt.title(f'Mean Filtered (MSE: {mse_mean:.2f})')
plt.axis("off")

plt.subplot(3, 2, 4)
plt.imshow(median_filtered, cmap='gray')
plt.title(f'Median Filtered (MSE: {mse_median:.2f})')
plt.axis("off")

plt.subplot(3,2,5)
plt.imshow(gaussian_filtered, cmap='gray')
plt.title(f'Gaussian Filtered (MSE: {mse_gaussian:.2f})')
plt.axis("off")
```
