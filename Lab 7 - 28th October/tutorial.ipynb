```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv2.imread(r"E:\SRM\Digital Image\Lab 7 - 28th October\Cameraman.png",0)
M,N = img.shape
```

```python
gauss_noise=np.zeros((M,N),dtype=np.uint8)
cv2.randn(gauss_noise,128,20)
gauss_noise=(gauss_noise*0.5).astype(np.uint8)
```

```python
combined=img+gauss_noise
```

```python
plt.subplot(1,3,1)
plt.imshow(img,cmap='gray')
plt.title('Original Image')
plt.axis("off")
plt.subplot(1,3,2)
plt.imshow(gauss_noise,cmap='gray')
plt.title('Gaussian Noise')
plt.subplot(1,3,3)
plt.imshow(combined,cmap='gray')
plt.title('combined')
plt.axis("off")
plt.show()
```

```python
# uni_noise=np.zeros((640,480),dtype=np.uint8)
uni_noise=np.zeros((M,N),dtype=np.uint8)
cv2.randu(uni_noise,0,255)
uni_noise=(uni_noise*0.5).astype(np.uint8)
plt.imshow(uni_noise,cmap='gray')
plt.axis("off")
plt.show()
```

```python
# imp_noise=np.zeros((640,480),dtype=np.uint8)
imp_noise=np.zeros((M,N),dtype=np.uint8)
cv2.randu(imp_noise,0,255)
imp_noise=cv2.threshold(imp_noise,200,255,cv2.THRESH_BINARY)[1]
plt.imshow(imp_noise,cmap='gray')
plt.axis("off")
plt.show()
```

```python
image2 = img + uni_noise
image3 = img + imp_noise
```

```python
plt.imshow(image2,cmap='gray')
plt.axis('off')
plt.show()

plt.imshow(image3,cmap='gray')
plt.axis('off')
plt.show()
```
