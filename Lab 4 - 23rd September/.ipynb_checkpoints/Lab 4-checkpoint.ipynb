### Histogram Equalization

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def histogram_equalization(gray_levels, nk):
   
    MN = np.sum(nk)
    pdf = nk / MN
    cdf = np.cumsum(pdf)
    L = len(gray_levels)
    Sk=(L - 1) * cdf
    Sk_roundOff = np.round(Sk).astype(int)

    new_histogram = np.zeros(L)
    for i in range(len(Sk)):
        new_histogram[Sk_roundOff[i]] += nk[i]

    df = pd.DataFrame({
        'GrayLevel': gray_levels,
        'nk': nk,
        'PDF (nk/MN)': pdf,
        'CDF': cdf,
        'Sk':np.round(Sk,2),
        'Sk (roundOff)': Sk_roundOff
    })

    df_new=pd.DataFrame({
        'GrayLevel': gray_levels,
        'new nk':new_histogram
    })
    pd.set_option('display.float_format', lambda x: '%.6f' % x)
    return df.to_string(index=False, justify='center'),df_new.to_string(index=False, justify='center'), new_histogram


gray_levels = np.array([0, 1, 2, 3, 4, 5, 6, 7])
nk = np.array([790, 1023, 850, 656, 329, 245, 122, 81])


df_result,df_new_result, new_hist = histogram_equalization(gray_levels, nk)


print(df_result)


fig, axs = plt.subplots(1, 2, figsize=(12, 5))


axs[0].bar(gray_levels, nk, color='red', alpha=0.7, edgecolor='black')
axs[0].set_xlabel('GrayLevel')
axs[0].set_ylabel('Frequency (nk)')
axs[0].set_title('Initial Histogram')
axs[0].set_xticks(gray_levels)
axs[0].grid(axis='y', linestyle='--', alpha=0.7)


axs[1].bar(gray_levels, new_hist, color='blue', alpha=0.7, edgecolor='black')
axs[1].set_xlabel('Gray Level (Sk)')
axs[1].set_ylabel('Frequency (new nk)')
axs[1].set_title('New Histogram After Equalization')
axs[1].set_xticks(gray_levels)
axs[1].grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
print("New Table")
print(df_new_result)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def histogram_equalization(gray_levels, nk):
    # Total number of pixels
    MN = np.sum(nk)

    # Step 1: Calculate PDF (nk/MN)
    pdf = nk / MN

    # Step 2: Calculate CDF (cumulative sum of PDF)
    cdf = np.cumsum(pdf)

    # Step 3: Calculate new intensity values (Sk = round((L-1) * CDF))
    L = len(gray_levels)
    Sk = np.round((L - 1) * cdf).astype(int)  # assuming L is the number of intensity levels (8 in this case)

    # Step 4: Create the resulting histogram with new gray levels
    new_histogram = np.zeros(L)
    for i in range(len(Sk)):
        new_histogram[Sk[i]] += nk[i]

    # Create a DataFrame to display the results
    df = pd.DataFrame({
        'Gray Level (rk)': gray_levels,
        'nk': nk,
        'PDF (nk/MN)': pdf,
        'CDF': cdf,
        'Sk (new gray level)': Sk
    })

    return df, new_histogram

# Example input based on the table in the image
gray_levels = np.array([0, 1, 2, 3, 4, 5, 6, 7])
nk = np.array([790, 1023, 850, 656, 329, 245, 122, 81])

# Perform histogram equalization
df_result, new_hist = histogram_equalization(gray_levels, nk)

# Display the table result
print(df_result)

# Plot the histogram of the new nk values against gray levels
plt.bar(gray_levels, new_hist, color='blue', alpha=0.7, edgecolor='black')
plt.xlabel('Gray Level (Sk)')
plt.ylabel('Frequency (nk)')
plt.title('New Histogram After Equalization')
plt.xticks(gray_levels)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
