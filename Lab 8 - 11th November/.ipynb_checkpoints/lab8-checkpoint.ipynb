### 1. Develop programs to implement the first derivative using Robert, Prewitt and Sobel ooperators. Take mask size as 3X3. Display images side by side..

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt 
```

```python
img = cv2.imread(r"E:\SRM\Digital Image\Lab 8 - 11th November\Cameraman.png",0)
```

```python
# Robert operator 
robertx = np.array([[-1,0],
                   [0,1]])
roberty = np.rot90(robertx)
# after convolution
robert_conx = cv2.filter2D(img,-1,robertx)
robert_cony = cv2.filter2D(img,-1,roberty)
robert_mag = np.sqrt(np.square(robert_conx)+np.square(robert_cony))
robert_direction = np.arctan2(robert_cony,robert_conx)
```

```python
# Prewitt operator 
Prewittx = np.array([[-1, -1, -1],
                      [ 0,  0,  0],
                      [ 1,  1,  1]])
Prewitty = np.rot90(Prewittx)
# after convolution
Prewitt_conx = cv2.filter2D(img,-1,Prewittx)
Prewitt_cony = cv2.filter2D(img,-1,Prewitty)
Prewitt_mag = np.sqrt(np.square(Prewitt_conx)+np.square(Prewitt_cony))
Prewitt_direction = np.arctan2(Prewitt_cony,Prewitt_conx)
```

```python
#Sobel
# sobelx= np.array([[-1,0,1],[-2,0,2],[-1,0,1]])
sobelx = np.array([[-1, -2, -1],
                   [ 0,  0,  0],
                   [ 1,  2,  1]])
sobely = np.rot90(sobelx)
#convoling the image using 3x3 kernal.Sobel horizontal and verticle to the input image to the image using the above kernal
sobelx_con = cv2.filter2D(img, -1, sobelx)
sobely_con = cv2.filter2D(img, -1, sobely)
magnitude_sobel = np.sqrt(np.square(sobelx_con) + np.square(sobely_con))
gradient_direction_sobel = np.arctan2(sobely_con, sobelx_con)
```

```python
plt.figure(figsize=(20, 10))
plt.subplot(1,5,1)
plt.title('Original Image')
plt.imshow(img,cmap='gray')
plt.axis('off')
plt.subplot(1,5,2)
plt.title('Robert Horizontal')
plt.imshow(robert_conx,cmap='gray')
plt.axis('off')
plt.subplot(1,5,3)
plt.title('Robert Vertical')
plt.imshow(robert_cony,cmap='gray')
plt.axis('off')
plt.subplot(1,5,4)
plt.title('Robert Magnitude')
plt.imshow(robert_mag,cmap='gray')
plt.axis('off')
plt.subplot(1,5,5)
plt.title('Robert direction')
plt.imshow(robert_direction,cmap='gray')
plt.axis('off')
plt.show()

plt.figure(figsize=(20, 10))
plt.subplot(2,5,1)
plt.title('Original Image')
plt.imshow(img,cmap='gray')
plt.axis('off')
plt.subplot(2,5,2)
plt.title('Prewitt Horizontal')
plt.imshow(Prewitt_conx,cmap='gray')
plt.axis('off')
plt.subplot(2,5,3)
plt.title('Prewitt Vertical')
plt.imshow(Prewitt_cony,cmap='gray')
plt.axis('off')
plt.subplot(2,5,4)
plt.title('Prewitt Magnitude')
plt.imshow(Prewitt_mag,cmap='gray')
plt.axis('off')
plt.subplot(2,5,5)
plt.title('Prewitt direction')
plt.imshow(Prewitt_direction,cmap='gray')
plt.axis('off')
plt.show()

plt.figure(figsize=(20, 10))
plt.subplot(3,5,1)
plt.title('Original Image')
plt.imshow(img,cmap='gray')
plt.axis('off')
plt.subplot(3,5,2)
plt.title('Sobel Horizontal')
plt.imshow(sobelx_con,cmap='gray')
plt.axis('off')
plt.subplot(3,5,3)
plt.title('Sobel Vertical')
plt.imshow(sobely_con,cmap='gray')
plt.axis('off')
plt.subplot(3,5,4)
plt.title('Sobel')
plt.imshow(magnitude_sobel,cmap='gray')
plt.axis('off')
plt.subplot(3,5,5)
plt.title('Sobel direction')
plt.imshow(gradient_direction_sobel,cmap='gray')
plt.axis('off')
plt.show()
```

### 3. Use Sobel operator for first derivative and apply LOG.

```python

# Sobel operator 
sobelx = np.array([[-1, -2, -1],
                   [ 0,  0,  0],
                   [ 1,  2,  1]])
sobely = np.rot90(sobelx)  

# Convolution with Sobel kernels
sobelx_con = cv2.filter2D(img, -1, sobelx)
sobely_con = cv2.filter2D(img, -1, sobely)
gradient_magnitude = np.sqrt(np.square(sobelx_con) + np.square(sobely_con))

# Apply Gaussian Blur (for Laplacian of Gaussian)
gaussian_blur = cv2.GaussianBlur(img, (5, 5), 0)

# Apply Laplacian to the smoothed image (second derivative)
laplacian = cv2.Laplacian(gaussian_blur, cv2.CV_64F)

# Normalize the gradient magnitude and Laplacian results for display
gradient_magnitude = np.uint8(np.clip(gradient_magnitude, 0, 255))
laplacian = np.uint8(np.clip(laplacian, 0, 255))

# Apply Log transformation (LOG) to the magnitude and Laplacian
log_gradient_magnitude = np.uint8(np.clip(np.log1p(gradient_magnitude), 0, 255))
log_laplacian = np.uint8(np.clip(np.log1p(np.abs(laplacian)), 0, 255))

# Plotting the results with the same structure
plt.figure(figsize=(20, 10))

# Original Image
plt.subplot(2, 3, 1)
plt.title('Original Image')
plt.imshow(img, cmap='gray')
plt.axis('off')

# Gradient Magnitude (Sobel)
plt.subplot(2, 3, 2)
plt.title('Gradient Magnitude (Sobel)')
plt.imshow(gradient_magnitude, cmap='gray')
plt.axis('off')

# Log of Gradient Magnitude (Sobel)
plt.subplot(2, 3, 3)
plt.title('Log of Gradient Magnitude (Sobel)')
plt.imshow(log_gradient_magnitude, cmap='gray')
plt.axis('off')

# Gaussian Blurred Image
plt.subplot(2, 3, 4)
plt.title('Gaussian Blurred Image')
plt.imshow(gaussian_blur, cmap='gray')
plt.axis('off')

# Laplacian of Gaussian
plt.subplot(2, 3, 5)
plt.title('Laplacian of Gaussian')
plt.imshow(laplacian, cmap='gray')
plt.axis('off')

# Log of Laplacian
plt.subplot(2, 3, 6)
plt.title('Log of Laplacian of Gaussian')
plt.imshow(log_laplacian, cmap='gray')
plt.axis('off')

# Show the plots
plt.show()
```

### 4. Demonstrate that LOG is equivalent to DOG, by applying both on the same image..

```python

# Define the LoG kernel (5x5)
log_kernel = np.array([[0, 0, -1, 0, 0], 
                       [0, -1, -2, -1, 0], 
                       [-1, -2, 16, -2, -1], 
                       [0, -1, -2, -1, 0], 
                       [0, 0, -1, 0, 0]], dtype=np.float32)

# Apply LoG filter
log_filtered_image = cv2.filter2D(img, -1, log_kernel)

# Apply DoG filter with different sigmas
sigma1 = 0.91
sigma2 = 1

gaussian1 = cv2.GaussianBlur(img, (0, 0), sigma1)
gaussian2 = cv2.GaussianBlur(img, (0, 0), sigma2)

dog_filtered_image = gaussian1 - gaussian2

# Convert results to uint8
log_filtered_image_uint8 = cv2.convertScaleAbs(log_filtered_image)
dog_filtered_image_uint8 = cv2.convertScaleAbs(dog_filtered_image)

# Plot the images
plt.figure(figsize=(20, 10))

# Original Image
plt.subplot(2, 3, 1)
plt.title('Original Image')
plt.imshow(img, cmap='gray')
plt.axis('off')

# LoG Filtered Image
plt.subplot(2, 3, 2)
plt.title('Laplacian of Gaussian (LoG)')
plt.imshow(log_filtered_image_uint8, cmap='gray')
plt.axis('off')

# DoG Filtered Image
plt.subplot(2, 3, 3)
plt.title('Difference of Gaussians (DoG)')
plt.imshow(dog_filtered_image_uint8, cmap='gray')
plt.axis('off')

plt.show()
```
