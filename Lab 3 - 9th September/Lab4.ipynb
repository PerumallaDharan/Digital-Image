```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv2.imread('E:\SRM\Digital Image\Lab 4 - 9th September\Cameraman.png')
plt.imshow(img)
plt.xticks([]),plt.yticks([])
plt.title('Original Image')
plt.show()
```

```python
blur_img=cv2.blur(img,[5,5])
```

```python
plt.subplot(1,2,1)
plt.imshow(img)
plt.xticks([]),plt.yticks([])
plt.title("Original Image")
plt.subplot(1,2,2)
plt.imshow(blur_img)
plt.xticks([]),plt.yticks([])
plt.title('Blur Image')
plt.show()
```

### Averaging Filter

```python
kernel=np.ones((5,5),dtype=float)/25
blur_kernel_img=cv2.filter2D(img,-1,kernel)
```

```python
plt.subplot(1,3,1)
plt.imshow(img)
plt.xticks([]),plt.yticks([])
plt.title("Original Image")
plt.subplot(1,3,2)
plt.imshow(blur_img)
plt.xticks([]),plt.yticks([])
plt.title('Blur Image')
plt.subplot(1,3,3)
plt.imshow(blur_kernel_img)
plt.xticks([]),plt.yticks([])
plt.title('Blur with kernel')
plt.show()
```

### Weighted Filter

```python
kernel_weighted=np.array([[1,2,1],[2,4,2],[1,2,1]])
weighted_blur=cv2.filter2D(img,-1,kernel_weighted)
```

```python
plt.subplot(2,3,1)
plt.imshow(img)
plt.xticks([]),plt.yticks([])
plt.title("Original Image")
plt.subplot(2,3,2)
plt.imshow(blur_img)
plt.xticks([]),plt.yticks([])
plt.title('Blur Image')
plt.subplot(2,3,3)
plt.imshow(blur_kernel_img)
plt.xticks([]),plt.yticks([])
plt.title('Blur with kernel')
plt.subplot(2,3,4)
plt.imshow(weighted_blur)
plt.xticks([]),plt.yticks([])
plt.title('Weighted Blur Image')
plt.show()
```

### Median Filter

```python
median_blur=cv2.medianBlur(img,5)
```

```python
plt.subplot(2,3,1)
plt.imshow(img)
plt.xticks([]),plt.yticks([])
plt.title("Original Image")
plt.subplot(2,3,2)
plt.imshow(blur_img)
plt.xticks([]),plt.yticks([])
plt.title('Blur Image')
plt.subplot(2,3,3)
plt.imshow(blur_kernel_img)
plt.xticks([]),plt.yticks([])
plt.title('Blur with kernel')
plt.subplot(2,3,4)
plt.imshow(weighted_blur)
plt.xticks([]),plt.yticks([])
plt.title('Weighted Blur Image')
plt.subplot(2,3,5)
plt.imshow(median_blur)
plt.xticks([]),plt.yticks([])
plt.title('Median Blur Image')
plt.show()
```

```python
ker1=np.ones((5,5),dtype=float)/25
blur_kernel_img1=cv2.filter2D(img,-1,ker1)

ker2=np.ones((17,17),dtype=float)/(17*17)
blur_kernel_img2=cv2.filter2D(img,-1,ker2)

ker3=np.ones((25,25),dtype=float)/(25*25)
blur_kernel_img3=cv2.filter2D(img,-1,ker3)

ker4=np.ones((35,35),dtype=float)/(35*35)
blur_kernel_img4=cv2.filter2D(img,-1,ker4)

ker5=np.ones((45,45),dtype=float)/(45*45)
blur_kernel_img5=cv2.filter2D(img,-1,ker5)
```

```python
plt.subplot(2,3,1)
plt.imshow(img)
plt.xticks([]),plt.yticks([])
plt.title("Original Image")

plt.subplot(2,3,2)
plt.imshow(blur_kernel_img1)
plt.xticks([]),plt.yticks([])
plt.title("5X5 Kernel")

plt.subplot(2,3,3)
plt.imshow(blur_kernel_img2)
plt.xticks([]),plt.yticks([])
plt.title("17X17 Kernel")

plt.subplot(2,3,4)
plt.imshow(blur_kernel_img3)
plt.xticks([]),plt.yticks([])
plt.title("25X25 Kernel")

plt.subplot(2,3,5)
plt.imshow(blur_kernel_img4)
plt.xticks([]),plt.yticks([])
plt.title("35X35 Kernel")

plt.subplot(2,3,6)
plt.imshow(blur_kernel_img5)
plt.xticks([]),plt.yticks([])
plt.title("45X45 Kernel")
plt.show()
```

```python
weighted_ker1=np.array([[1,2,1],[2,4,2],[1,2,1]])
weighted_blur1=cv2.filter2D(img,-1,weighted_ker1)

weighted_ker2=np.array([[1,2,1,2,1],[2,4,2,4,2],[1,2,1,2,1],[2,4,2,4,2],[1,2,1,2,1]])
weighted_blur2=(cv2.filter2D(img,-1,weighted_ker2))

weighted_ker3=np.array([[1,2,1,2,1,2,1],[2,4,2,4,2,4,2],[1,2,1,2,1,2,1],[2,4,2,4,2,4,2],[1,2,1,2,1,2,1],[2,4,2,4,2,4,2],[1,2,1,2,1,2,1]])
weighted_blur3=(cv2.filter2D(img,-1,weighted_ker3))

weighted_ker_std=np.array([[1,4,7,4,1],[4,16,26,16,4],[7,26,41,26,7],[4,16,26,16,4],[1,4,7,4,1]])/273
weighted_blur_std=(cv2.filter2D(img,-1,weighted_ker_std))

weighted_ker_std2=np.array([[0,0,1,2,1,0,0],[0,3,13,22,13,3,0],[1,13,59,97,59,13,1],[2,22,97,159,97,22,2],[1,13,59,97,59,13,1],[0,3,13,22,13,3,0],[0,0,1,2,1,0,0]])/1003
weighted_blur_std2=(cv2.filter2D(img,-1,weighted_ker_std2))
```

```python
plt.subplot(3,2,1)
plt.imshow(img)
plt.xticks([]),plt.yticks([])
plt.title("Original Image")

plt.subplot(3,2,2)
plt.imshow(weighted_blur1)
plt.xticks([]),plt.yticks([])
plt.title("3X3 Weighted Kernel")

plt.subplot(3,2,3)
plt.imshow(weighted_blur2)
plt.xticks([]),plt.yticks([])
plt.title("5X5 Weighted Kernel")

plt.subplot(3,2,4)
plt.imshow(weighted_blur3)
plt.xticks([]),plt.yticks([])
plt.title("7X7 Weighted Kernel")

plt.subplot(3,2,5)
plt.imshow(weighted_blur_std)
plt.xticks([]),plt.yticks([])
plt.title("5X5 Weighted Kernel Standard")

plt.subplot(3,2,6)
plt.imshow(weighted_blur_std2)
plt.xticks([]),plt.yticks([])
plt.title("7X7 Weighted Kernel Standard")

plt.show()
```
