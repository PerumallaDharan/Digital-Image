# Lab-02 Intensity Transformation

##### 1. Develop programs for following spatial filtering operations on a gray scale image.


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv2.imread('D:/D.I.P/cameraman.jpeg')
plt.imshow(img)
plt.xticks([]),plt.yticks([])
plt.title("Original Image")
plt.show()
```

```python
blur_img=cv2.blur(img,[5,5])
```

```python
plt.subplot(1,2,1)
plt.imshow(img)
plt.xticks([]),plt.yticks([])
plt.title("Original Image")
plt.subplot(1,2,2)
plt.imshow(blur_img)
plt.xticks([]),plt.yticks([])
plt.title('Blur Image')
plt.show()
```

### Averaging Filter

```python
kernel=np.ones((5,5),dtype=float)/25
blur_kernel_img=cv2.filter2D(img,-1,kernel)
```

```python
plt.subplot(1,3,1)
plt.imshow(img)
plt.xticks([]),plt.yticks([])
plt.title("Original Image")
plt.subplot(1,3,2)
plt.imshow(blur_img)
plt.xticks([]),plt.yticks([])
plt.title('Blur Image')
plt.subplot(1,3,3)
plt.imshow(blur_kernel_img)
plt.xticks([]),plt.yticks([])
plt.title('Blur with kernel')
plt.show()
```

### Weighted Filter

```python
kernel_weighted=np.array([[1,2,1],[2,4,2],[1,2,1]])
weighted_blur=cv2.filter2D(img,-1,kernel_weighted)
```

```python
plt.subplot(2,3,1)
plt.imshow(img)
plt.xticks([]),plt.yticks([])
plt.title("Original Image")
plt.subplot(2,3,2)
plt.imshow(blur_img)
plt.xticks([]),plt.yticks([])
plt.title('Blur Image')
plt.subplot(2,3,3)
plt.imshow(blur_kernel_img)
plt.xticks([]),plt.yticks([])
plt.title('Blur with kernel')
plt.subplot(2,3,4)
plt.imshow(weighted_blur)
plt.xticks([]),plt.yticks([])
plt.title('Weighted Blur Image')
plt.show()
```

### Median Filter

```python
median_blur=cv2.medianBlur(img,5)
```

```python
plt.subplot(2,3,1)
plt.imshow(img)
plt.xticks([]),plt.yticks([])
plt.title("Original Image")
plt.subplot(2,3,2)
plt.imshow(blur_img)
plt.xticks([]),plt.yticks([])
plt.title('Blur Image')
plt.subplot(2,3,3)
plt.imshow(blur_kernel_img)
plt.xticks([]),plt.yticks([])
plt.title('Blur with kernel')
plt.subplot(2,3,4)
plt.imshow(weighted_blur)
plt.xticks([]),plt.yticks([])
plt.title('Weighted Blur Image')
plt.subplot(2,3,5)
plt.imshow(median_blur)
plt.xticks([]),plt.yticks([])
plt.title('Median Blur Image')
plt.show()
```
